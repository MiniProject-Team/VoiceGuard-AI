# Phase 5 — Explainable risk engine
All values below are simulations, not measured security performance. The engine consumes structured Phase 3/4 outputs and never raw audio.

```python
import yaml
from src.risk import RiskEngine, RiskInput
config=yaml.safe_load(open('configs/risk_engine.yaml'))
engine=RiskEngine(config)
```

```python
genuine=RiskInput(synthetic_probability=.05,speaker_similarity=.9,speaker_verified=True,call_origin='known',contact_match=True,transaction_amount=1000)
clone=RiskInput(synthetic_probability=.95,speaker_similarity=.92,speaker_verified=True,call_origin='unknown',transaction_amount=250000,privileged_action=True)
missing=RiskInput(synthetic_probability=.3)
for item in [genuine,clone,missing]:
    result=engine.evaluate(item)
    display({k:result[k] for k in ['risk_score','risk_level','reasons','signal_breakdown','missing_signals','temporal']})
```

```python
temporal_engine=RiskEngine(config)
[temporal_engine.evaluate(clone)['temporal'] for _ in range(4)]
```

A high speaker match plus high synthetic probability is treated as a clone-risk pattern. `BLOCK_OR_ESCALATE` applies to authorization of sensitive actions; it does not terminate calls.
